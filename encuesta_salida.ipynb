```python
import pandas as pd
```

```python
archivo = "entrevistas_salida.xlsx"
df = pd.read_excel(archivo, sheet_name = "Respuestas de formulario")
```

```python
col_sug = [c for c in df.columns if c.startswith("¿Qué sugieres para mejorar las condiciones laborales de la empresa?")][0]
col_area = [c for c in df.columns if c.startswith("¿Área del restaurante donde desarrollaba sus actividades?")][0]
col_num = [c for c in df.columns if c.startswith("Numero de empleado (tal y como aparece en el siete Compaq)")][0]
```

```python
for i, s in df[col_sug].dropna().items():
    print(i, "|", s)
```

```python
import re

nombres = ["Nombre 1", "Nombre 2", "Nombre 3"]
patron = r"\b(" + "|".join(nombres) + r")\b"
```

```python
df[col_sug]= df[col_sug].str.replace(patron, "[nombre]", regex=True, flags= re.IGNORECASE)
df["Nombre"] = [f"Empleado {i}" for i in range(1, len(df)+1)]
df[col_num] = range(1, len(df)+1)
df[col_area] = df[col_area].replace({"Panadería" : "Otras áreas", "Mantenimiento": "Otras áreas"})

df.to_excel("entrevista_salida_anonima.xlsx", sheet_name="Respuestas de formulario", index=False)
```

```python
rev= pd.read_excel("entrevista_salida_anonima.xlsx")
print("Nombres que queda:", rev[col_sug].str.contains(patron, case=False, regex=True).sum())
print(rev[col_area].value_counts())
print(rev["Nombre"].head())
```
